````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.

````python
# База
cities_clean = list(map(lambda x: x.strip().title(), cities))
print(cities_clean)
amounts_int = list(map(int, amounts))
print(amounts_int)
orders_count, revenue, avg_revenue = len(amounts), round(sum(amounts_int)), round(sum(amounts_int) / len(amounts_int))
print(orders_count, revenue, avg_revenue)
not_paid = sum(not p for p in paid)
print(not_paid)
not_paid_sum = 0
for a, p in zip(amounts_int, paid):
  if not p:
    not_paid_sum += a
print(not_paid_sum)
for city, sum1, st in zip(cities_clean, amounts_int, paid):
  print(f'{city} | {sum1} | {st}')

````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
rev_ala = 0
rev_ast = 0
for sum1, city, st in zip(amounts_int, cities_clean, paid):
  if (city == 'Almaty') and (st):
      rev_ala = rev_ala + sum1
  elif (city == 'Astana') and (st):
      rev_ast = rev_ast + sum1
print(f'Выручка по Алмате : {rev_ala}, выручка по Астане : {rev_ast}')
idx_max = -1
for i in range(len(amounts_int)):
    if paid[i] and (idx_max == -1 or amounts_int[i] > amounts_int[idx_max]):
      idx_max = i
print(f'Индекс : {idx_max}, город : {cities_clean[idx_max]}')
week_days_sum = sum(day in ('сб', 'вс') for day in days)
print(f'Доля заказов выходных: {week_days_sum/len(days) * 100}')
#1
a = 0
month = 0
while a < 400_000:
  a = a + 15_000
  month += 1
print(f'За {month} месяцев студент накопит 400_000 откладывая 15_000')
#2
a = 0
dep = 15_000
month2 = 0
while a < 400_000:
  a = dep + a
  dep = 1000 + dep
  month2 += 1
print(f'За {month2} месяцев студент накопит 400_000 откладывая 15_000 и на 1000 больше чем в прошлый')




````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
unique_cities = []
for i in cities_clean:
  if i not in unique_cities:
    unique_cities.append(i)
revenue_by_city = [0] * len(unique_cities)
for sum1, city, st in zip(amounts_int, cities_clean, paid):
  if st:
    idx = unique_cities.index(city)
    revenue_by_city[idx] += sum1
sorted_unique_citites = [x[0] for x in sorted(list(zip(unique_cities, revenue_by_city)), key=lambda p: p[1], reverse=True)]
sorted_revenue_by_city = [x[1] for x in sorted(list(zip(unique_cities, revenue_by_city)), key=lambda p: p[1], reverse=True)]
print(sorted_unique_citites)
print(sorted_revenue_by_city)
s = sorted(amounts_int)
if len(amounts_int) % 2 == 0:
    med = (s[int(len(s)/2) - 1] + s[int(len(s)/2)])/2
else:
    med = s[int(len(s) - 1)//2]
print(med)
unique_days = []
for d in days:
    if d not in unique_days:
        unique_days.append(d)

revenue_by_day = [0] * len(unique_days)
for d, amount in zip(days, amounts_int):
    idx = unique_days.index(d)
    revenue_by_day[idx] += amount
for d, rev in zip(unique_days, revenue_by_day):
    print(f'{d} | {"█" * (rev // 5000)} {rev}')
````
